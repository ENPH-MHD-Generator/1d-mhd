# Recreate Freidberg

Work backwards using the equations in his paper to find some plasma properties to check the solver's accuracy

First, we'll define $\Delta T$, $\beta$, and $M$ as they are described in (7.3).

```python
delta_t = 8.783 - 1
beta = 8.249
M = 1.8
```

I'll use SymPy so solve for $Z$ in the energy balance equation (3.14).

```python
import sympy as sp

Z = sp.symbols('Z')

delta_t_eq = (5 / 9 * M**2)*(beta**2 * (beta**2 + (1 + Z)**2)) / (beta**2 + 1 + Z)**2 - delta_t
Z_val = max(sp.solve(delta_t_eq, Z))  # Take the positive root
print(Z_val)
```

We can also solve for $\eta$ using the definition of the Hall parameter.

```python
from scipy import constants

n_e = 7.478e19
B = 8.0
eta = sp.symbols(r"\eta")

eta_expr = B / (constants.elementary_charge * n_e * eta) - beta
eta_val = sp.solve(eta_expr, eta)[0]  # Only one value, just take it
print(eta_val)
```

We can also solve for the collision frequency from the definition of $\eta$ in (3.4).

```python
nu_M = sp.symbols('u')

nu_M_expr = constants.electron_mass * nu_M / (constants.elementary_charge**2 * n_e) - eta_val
nu_M_val = sp.solve(nu_M_expr, nu_M)[0]
print(nu_M_val / 1e9)
```

Which we can then use $\nu_M$ to solve for $Q$/$\sigma_{ep}$ from the definition of $\nu_M$ in (3.8).

```python
import math

Q = sp.symbols('Q')

n_p = n_e / 6.180e-7
Te = 4223
Q_expr = n_p * Q * math.sqrt(2 * constants.k * Te / constants.electron_mass) - nu_M_val
Q_val = sp.solve(Q_expr, Q)[0]
print(Q_val)
```

We can also find the load resistivity,

```python
eta_L = Z_val * eta_val
print(eta_L)
```
